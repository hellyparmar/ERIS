## 1. Setup & Dependencies

```python
import requests
import json
import pandas as pd
from datetime import datetime, timedelta
from pprint import pprint

# Base API URL
API_URL = "http://localhost:8000"

# Session for maintaining state
session = requests.Session()

print("✅ Dependencies loaded")
print(f"API Base URL: {API_URL}")
```

## 2. Health Check

```python
# Test API health
try:
    response = requests.get(f"{API_URL}/health")
    health_data = response.json()
    print("✅ API Health Status:")
    pprint(health_data)
except Exception as e:
    print(f"❌ Health check failed: {e}")
```

## 3. Authentication

```python
# Login with test credentials
email = "admin@rdios.com"
password = "admin123"

login_data = {
    'username': email,
    'password': password
}

try:
    response = requests.post(
        f"{API_URL}/api/v1/auth/login",
        data=login_data,
        headers={'Content-Type': 'application/x-www-form-urlencoded'}
    )
    auth_response = response.json()
    
    access_token = auth_response['access_token']
    user_data = auth_response['user']
    
    # Store in session headers
    session.headers.update({'Authorization': f'Bearer {access_token}'})
    
    print(f"✅ Logged in as: {user_data['full_name']} ({user_data['role']})")
    print(f"   Email: {user_data['email']}")
    print(f"   Token: {access_token[:50]}...")
except Exception as e:
    print(f"❌ Login failed: {e}")
```

## 4. Inventory Endpoints

```python
# List all inventory items
try:
    response = session.get(f"{API_URL}/api/v1/inventory")
    if response.status_code == 200:
        inventory = response.json()
        print(f"✅ Inventory items retrieved: {len(inventory) if isinstance(inventory, list) else 'N/A'}")
        print(json.dumps(inventory[:2] if isinstance(inventory, list) else inventory, indent=2))
    else:
        print(f"Status: {response.status_code}")
        print(response.text)
except Exception as e:
    print(f"❌ Error: {e}")
```

## 6. Available API Endpoints

```python
endpoints = {
    "System": [
        "GET /health - Health check",
        "GET / - Root endpoint"
    ],
    "Authentication": [
        "POST /api/v1/auth/login - User login",
        "POST /api/v1/auth/register - User registration",
        "POST /api/v1/auth/refresh - Refresh token"
    ],
    "Inventory": [
        "GET /api/v1/inventory - List all inventory",
        "POST /api/v1/inventory - Create inventory item",
        "GET /api/v1/inventory/{id} - Get inventory details"
    ],
    "Community": [
        "GET /api/v1/community - Community listings",
        "POST /api/v1/community - Create listing"
    ],
    "Invoices": [
        "GET /api/v1/invoices - List invoices",
        "POST /api/v1/invoices - Create invoice"
    ]
}

print("🔌 Available API Endpoints:\n")
for category, methods in endpoints.items():
    print(f"**{category}:**")
    for method in methods:
        print(f"  • {method}")
    print()
```
